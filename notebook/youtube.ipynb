```python
def get_transcript(url):

    video_id = get_video_id(url)

    api = YouTubeTranscriptApi()
    transcript = api.fetch(video_id, languages=["en", "hi"])

    segments = []

    for item in transcript:
        segments.append({
            "text": item.text,
            "start": round(item.start, 2)
        })

    return segments
```
